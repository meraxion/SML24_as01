## Model formulation

You can check for more details lectures on sampling methods and approximate methods. 

Overall, the vector/matrix representation of the data in this problem is given by:
$$
    x_{n}=\begin{pmatrix}x_{n1} \\ x_{n2}\end{pmatrix},  \quad   w=\begin{pmatrix}w_{1} \\ w_{2}\end{pmatrix},  \quad  X=\begin{pmatrix}x_{1}^{T} \\ x_{2}^{T} \\ \vdots \\ x_{N}^{T}\end{pmatrix}.
 $$   

The likelihood:

$$
p(t|X, w) = \prod_{n=1}^{N} P(T_{n}=1|x_{n},w)^{t_{n}} P(T_{n}=0|x_{n}, w)^{1-t_{n}}
     = \prod_{n=1}^{N}\left(\frac{1}{1+\exp(-w^{T}x_{n})} \right)^{t_{n}}\left(\frac{\exp(-w^{T}x_{n})}{1+\exp(-w^{T}x_{n})} \right)^{1-t_{n}}. 
$$    

We assume that the prior is Gaussian, i.e

$$
p(w)=N(0, \sigma^{2}\mathbb{I}),
$$
where $\sigma^{2} = 0.5$.

In this problem, we are interested in finding the posterior distribution of the weights

$$
p(w|t,X) = \frac{p(t|X,w)p(w)}{p(t|X)},
$$

and using it to make the response (class) predictions for a new object $x_{new}$, i.e. we are interested in the following expectation with respect to the posterior


$$
P(t_{new}=1|x_{new}, X, t) = \mathbb{E}_{p(w|X, t, \sigma^{2})}\left\{\frac{1}{1+\exp (-w^{T}x_{new})} \right\}. 
$$

Unfortunately, the posterior distribution and the expectation of the interest with respect to the posterior are not analytically tractable. Thus, we have to use one of the other techniques to evaluate these (MAP, Laplace approximation or sampling methods).

## Part 1  - The inference with MAP solution 

Let us denote $g(w; X, t, \sigma^{2}) = p(t|X,w)p(w)$ which corresponds to denominator of $p(w|t,X)$. As $g(w; X, t, \sigma^{2})$ is proportional to $p(w|t,X)$, the maximum of $g(\cdot)$ will also correspond to the maximum of $p(w|t,X)$. This is much easier since $p(t|X)$ is hard to evaluate. Thus, we can find the maximum a posterior estimate $\hat{w}$ by finding the maximum of $g(w; X, t, \sigma^{2})$. For that, we will use the Newton-Raphson algorithm. 

### The Newton-Raphson algorithm

The Newton-Raphson algorithm for optimization is similar to gradient descent. But unlike with gradient descent, the Newton-Rasphon algorithm also takes into information about the second-order derivatives. When the second derivative is available/easy to compute, the Newton-Raphson algorithm may be preferred due to faster convergence compared to gradient descent. The iteration rule of the Newton-Raphson algorithm reads:
$$
x_{n+1} = x_{n} - \frac{f'(x_{n})}{f''(x_{n})}.
$$

Recall for the binary response model, we have 

$$
\log g(w; X, t) = \sum_{n=1}^{N} \log P(T_{n}=t_{n}|x_{n}, w) +\log p(w|\sigma^{2})\\
= \sum_{n=1}^{N} \log \left[ \left( \frac{1}{1+\exp(-w^{T}x_{n})} \right)^{t_{n}} \left( \frac{\exp(-w^{T}x_{n})}{1+\exp (-w^{T}x_{n})}\right)^{1-t_{n}} \right]\\
+ \log p(w|\sigma^{2})
$$

For simplicity of notation, let us denote 
$$
P_{n} = P(T_{n}=1=w, x_{n}) = \frac{1}{1+\exp(-w^{T}x_{n})}. 
$$
Thus we get (assume that $w$ is D-dimensional):
$$
\log g(w; X, t) = \log p(w|\sigma^{2}) + \sum_{n=1}^{N} \log P_{n}^{t_{n}} + \log (1-P_{n})^{1-t_{n}}\\
=-\frac{D}{2}\log 2\pi -D \log \sigma - \frac{1}{2\sigma}w^{T}w + \sum_{n=1}^{N}\log P_{n} + (1-t_{n})\log(1-P_{n}).
$$
Then for the binary response model, the iteration rule of the Newton-Raphson algorithm is
$$
w_{i+1} = w_{i} -\left(\frac{\partial \log g(w:X,t)}{\partial w \partial w^{T}} \right)^{-1} \frac{\log g(w; X, t)}{\partial w}
$$

## Part 2 -- Laplace approximation

$$
\log g(w; X, t, \sigma^{2})\approx \log g(\hat{w}; X, t, \sigma^{2})-\frac{v}{2}(w-\hat{w}), 
$$
where 
$$
v = - \frac{\partial^{2}\log g(w;X, t, \sigma^{2})}{\partial w \partial w^{T}} \bigg|_{\hat{w}}.

## Part 3 - Metropolis-Hastings algorithm for the binary response problem

Generally, the formula for the acceptance probability in the Metropolis-Hastings algorithm is given by:

$$
\alpha(w^{c}, w^{(i)})=\min \left(1, \frac{p(w^{c})}{p(w^{(i)})} \frac{q(w^{(i)}|w^{c})}{q(w^{c}|w^{(i)})}\right),  
$$
where $p(w)$ is our target distribution. 

In this case, it would be more appropriate to use a full notation of $p(w|t, X, \mu_{0}, \sigma^{2}_{0})$ since our target function is the posterior distribution of $w$. Recall that the posterior is equal to the likelihood times the prior. 1. Write down the explicit formula for $p(w^{c})$ and the distribution for $q(w^{(i)}|w^{c})$ in the acceptance probability formula. Is the distribution $q(w^{(i)}|w^{c})$ symmetric?

**Begin Answer**

**Part 1 Target Distribution**

The target distribution is the posterior defined by the following:
$$
p(w^c|t,X) \propto p(t|X,w^c)p(w^c),
$$

Where

$$
\begin{align}
p(t|X,w^c) &= \prod_{n=1}^{N}\left(\frac{1}{1+\exp(-w^{T}x_{n})} \right)^{t_{n}}\left(\frac{\exp(-w^{T}x_{n})}{1+\exp(-w^{T}x_{n})} \right)^{1-t_{n}}\\

p(w^c) &= \mathcal{N}(0, \sigma^2 \mathbb{I}) = \frac{1}{2\pi\sigma^2}\exp\left[-\frac{w^Tw}{2\sigma^2}\right]
\end{align}
$$

Thus

$$
\begin{align}
p(w^c|t,X) \propto \underbrace{\frac{1}{2\pi\sigma^2}\exp\left[-\frac{w^Tw}{2\sigma^2}\right]}_\text{prior}\times \underbrace{\prod_{n=1}^{N}\left(\frac{1}{1+\exp(-w^{T}x_{n})} \right)^{t_{n}}\left(\frac{\exp(-w^{T}x_{n})}{1+\exp(-w^{T}x_{n})} \right)^{1-t_{n}}}_\text{likelihood}
\end{align}
$$

To evaluate this expression, it is simpler to take the log of the expression in two stages. Lets begin with the prior

$$
\begin{align}
\log{p(w^c)} = \log{\left[\frac{1}{2\pi\sigma^2}\exp\left[-\frac{w^Tw}{2\sigma^2}\right]\right]} = -\frac{w^Tw}{2\sigma^2} -\log{2\pi\sigma^2}
\end{align}
$$

Lets do the likelihood now:

$$
\begin{align}
\log p(t|X,w^c) &= \sum_{n=1}^N\left(t_n[-\log(1+\exp(-w^Tx_n))] + (1-t_n)\left[(-w^Tx_n)-\log(1+\exp(-w^Tx_n))\right]\right)\\
                 &= \sum_{n=1}^N\left(-t_n\log(1+\exp(-w^Tx_n))-w^Tx_n-\log(1+\exp(-w^Tx_n))+t_nw^Tx_n+t_n\log(1+\exp(-w^Tx_n))\right)\\
                 &= \sum_{n=1}^N\left(w^Tx_n(t_n-1)-\log(1+\exp(-w^Tx_n))\right)
\end{align}
$$

Thus for posterior we have:

$$
\log p(w^c|t,X) \propto \underbrace{-\frac{w^Tw}{2\sigma^2} -\log{2\pi\sigma^2}}_\text{prior} + \underbrace{\sum_{n=1}^N\left(w^Tx_n(t_n-1)-\log(1+\exp(-w^Tx_n))\right)}_\text{likelihood} 
$$

**Part 2 Proposal Distribution**
$$
q(w^{(i)}|w^{c}) = \mathcal{N}(0, 0.5\mathbb{I}) = \frac{1}{2\pi\sigma^2}\exp\left[-\frac{(w^c)^Tw^c}{2\sigma^2}\right] = \frac{1}{\pi}\exp\left[-(w^c)^Tw^c\right]
$$

Given that the proposal distribution is a normal distribution, it is symmetric.

2. Using the information from the previous question (i.e., $p(\cdot)$ and the proposal distribution $q(\cdot|\cdot)$), create a function `metropolis_hastings()`. 

```python
def log_joint_probability(x, t, w):
    """
    Calculates log of joint probability.

    Parameters
    ----------
    x : (num_data, 2) np.ndarray
        The attributes, where num_data is our sample size
    t : (num_data,) np.ndarray
        Class labels
    w : (1, 2) np.ndarray
        Weights.

    Returns
    -------
    log-probability of w
    """
    x = x[:,np.newaxis]
    t = t[:,np.newaxis]
    # YOUR CODE HERE
    first_term = np.sum(x@w.T*(t-1)-np.log(1+np.exp(-x@w)))
    second_term = -w.T@w

    return first_term + second_term
    
```

```python
def acceptance_function(x, t, w_candidate, w_current):
    """
    Computes acceptance probability of a candidate.

    Parameters
    ----------
    X : (N, 2) np.ndarray
        Data points.
    rho_candidate : float
        Candidate sample.
    rho_current : float
        Current sample.

    Returns
    -------
    Value of acceptance function for given rho_candidate and rho_current.
    """
    numerator = log_joint_probability(x, t, w_candidate)
    denominator = log_joint_probability(x, t, w_current)

    return min(1, np.exp(numerator-denominator))
```

```python
from scipy.stats import multivariate_normal
def metropolis_hastings(n_samps, jump_sigma, x, t):
    """
    Metropolis-Hastings algorithm

    Parameters
    ----------
    n_samps: integer
        Number of Metropolis-Hastings samples
    jump_sigma: positive scalar
        Standard deviation of the proposal distribution (we assume it is the same for both parameters)
    x : (num_data, 2) np.ndarray
        The attributes, where num_data is our sample size
    t : (num_data,) np.ndarray
        Class labels
    Returns
    -------
    w_samples (n_samps, 2) np.ndarray
              Metropolis-Hastings samples for (w1, w2)
    
    """
    accepted = 0
    w_samples = np.zeros((n_samps,2))

    proposal_mean = np.array([0, 0])
    proposal_cov = jump_sigma*np.eye(2)
    proposal = multivariate_normal(proposal_mean, proposal_cov)

    for i in range(0, n_samps-1):
        # Draw sample from proposal distribution
        w_candidate = proposal.rvs(1)
        acceptance_rate = acceptance_function(x, t, w_candidate, w_samples[i])

        if acceptance_rate > np.random.uniform(0,1,1):
            w_samples[i+1] = w_candidate
            accepted += 1
        else:
            w_samples[i+1] = w_samples[i]

    acceptance_rate = accepted/n_samps

    return w_samples
```

5. Comment on the results. How do decision probabilities differ for MAP estimation, Laplace approximation and Metropolis-Hastings? What are the strengths and weaknesses of each of these methods?

Comparison: All three methods are a way of approximating the posterior distribution when it is not analytically tractable. 

**MAP Estimation with Newton Raphson**: This method is similar to gradient descent but it also considers second order derivative information which means it is usually faster than gradient descent (if the second order derivatives are easy to calculate). The MAP approximation only takes about 4 iterations to converge!
It also provides a single best estimate for the parameters and it is computationally cheap since it reduces the problem to optimization. However some disadvantages are that it does not provide any uncertainty about the estimates, and it might get stuck in a local optima. 

**Laplace Approximation**: Slightly more flexible than the MAP estimate - it assumes the posterior can be modelled as a guassian with a mean at the MAP estimate. This captures some uncertainty, but makes assumptions about the form of the posterior (that it is a unimodal Guassian), so it may not be appropriate for very complex distributions that are multi-modal. However it is cheaper than full sampling methods.

**Metropolis Hastings**: This algorithm is ideal if we have access to the likelihood and can compute it. A disadvantage is that it can take many samples to come to an optimal solution. The decision probability is similar to that of the Laplace approximation - the lines "fan out" as we move farther away from the center of the data points. Compare this to the MAP estimation, which does not demonstrate the fanning behavior. This method is good for when we have a complex posterior, as it makes no assumptions about the form of the posterior.
